```python
from pathlib import Path
import pandas as pd

# Resolve the raw data directory from common notebook working directories.
candidate_dirs = [
    Path("data/clean"),
    Path("../data/clean"),
    Path.cwd() / "data" / "clean",
    Path.cwd().parent / "data" / "clean",
]

raw_dir = next((p.resolve() for p in candidate_dirs if p.exists()), None)
if raw_dir is None:
    raise FileNotFoundError("Could not find data/clean directory from the current notebook location.")

files = sorted(
    p for p in raw_dir.iterdir()
    if p.is_file() and (p.name.endswith(".csv") )
)

if not files:
    print(f"No .tsv or .tsv.zip files found in {raw_dir}")

for file_path in files:
    try:
        df = pd.read_csv(
            file_path,
            sep="\t",
            nrows=10,
            dtype=str,
            low_memory=False,
            compression="infer",
        )
        print(f"\n{file_path.name}")
        print(list(df.columns))
        print( df.duplicated().sum())
    except Exception as exc:
        print(f"\n{file_path.name}")
        print(f"Error reading file: {exc}")
```
